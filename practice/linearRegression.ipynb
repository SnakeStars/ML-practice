```python
import numpy as np
import matplotlib.pyplot as plt

class LinearRegression():

    def fit(self, X, y):
        rng = np.random.default_rng()
        self.weights = rng.random((X.shape[1],))
        steps = 100
        alpha = 0.1
        for step in range(steps):
            self.weights -= alpha * (2/X.shape[0]) * X.T @ (X @ self.weights - y)

        return self

    def predict(self, X):
        return X @ self.weights
```

```python
rng = np.random.default_rng(42)
X = rng.uniform(-3, 3, size=(50,1))
y = 2 * X[:,0] + rng.normal(0, 0.7, size = 50)

model = LinearRegression().fit(X,y)

grid = np.linspace(-3, 3, 100).reshape(-1, 1)
preds = model.predict(grid)

plt.scatter(X[:,0], y)
plt.plot(grid[:,0], preds, color="red")
plt.show()
```
